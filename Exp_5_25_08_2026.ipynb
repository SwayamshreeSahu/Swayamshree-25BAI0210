```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
df = pd.read_csv("/content/Iris.csv")
display(df.head())
```

```python
print("\nDataset Info:")
df.info()
print("\nDescriptive Statistics:")
display(df.describe())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
sns.histplot(df['SepalLengthCm'], kde=True, ax=axes[0,0], color='teal')
axes[0,0].set_title('Distribution of Sepal Length (Cm)')
sns.histplot(df['SepalWidthCm'], kde=True, ax=axes[0,1], color='darkorange')
axes[0,1].set_title('Distribution of Sepal Width (Cm)')
sns.histplot(df['PetalLengthCm'], kde=True, ax=axes[1,0], color='purple')
axes[1,0].set_title('Distribution of Petal Length (Cm)')
sns.histplot(df['PetalWidthCm'], kde=True, ax=axes[1,1], color='brown')
axes[1,1].set_title('Distribution of Petal Width (Cm)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Species', data=df, palette='magma')
plt.title('Distribution of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
sns.pairplot(df.drop('Id', axis=1), hue='Species', palette='tab10')
plt.suptitle('Pair Plot of Iris Features by Species', y=1.02) # Adjust suptitle position
plt.show()
```

```python
correlation_matrix = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='viridis', fmt=".2f")
plt.title('Correlation Matrix of Iris Features')
plt.show()
```
